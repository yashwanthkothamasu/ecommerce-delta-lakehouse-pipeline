```python
#changing the status of records
from pyspark.sql.functions import lit
df_updates = df_cdc_orders.withColumn("order_status",lit("cancelled")).limit(50)
```

```python
#create some new records
from pyspark.sql.functions import concat,col
df_new = df_cdc_orders.withColumn("order_id", concat(col("order_id") , lit("_new")))\
                      .withColumn("order_status",lit("created")).limit(20)
```

```python
#union 
df_union = df_updates.union(df_new)
```

```python
#writing to a staging table
df_union.write.format("delta").mode("overwrite").saveAsTable("ecommerce.bronze.cdc_batch")
```

```python
%sql
--MERGE INTO( SCD Type 1 : OVERWRITE)
MERGE INTO ecommerce.silver.orders as target
USING ecommerce.bronze.cdc_batch as source
on source.order_id = target.order_id

WHEN MATCHED THEN
UPDATE SET
target.order_status = source.order_status,
target.order_approved_at = source.order_approved_at,
target.order_delivered_carrier_date = source.order_delivered_carrier_date,
target.order_delivered_customer_date = source.order_delivered_customer_date

WHEN NOT MATCHED THEN 
INSERT(target.order_id,target.customer_id,target.order_status,target.order_purchase_timestamp,target.order_approved_at,target.order_delivered_carrier_date,target.order_delivered_customer_date,target.order_estimated_delivery_date,target._rescued_data)  values (source.order_id,source.customer_id,source.order_status,source.order_purchase_timestamp,source.order_approved_at,source.order_delivered_carrier_date,source.order_delivered_customer_date,source.order_estimated_delivery_date,source._rescued_data)
```

```python
%sql
-- Daily Revenue by Category

create or replace table ecommerce.gold.daily_revenues 
as 
select to_date(order_purchase_timestamp) as order_date,
p.product_category_name,
sum(oi.price) as total_revenue,
sum(oi.freight_value) as total_freight,
avg(oi.price) as avg_order_value,
count(DISTINCT o.order_id) as total_orders

from ecommerce.silver.orders o 
join ecommerce.silver.orderitems oi on o.order_id = oi.order_id
join ecommerce.gold.dim_products p on oi.product_id = p.product_id
where o.order_status = 'delivered'
group by 1,2;
```

```python
%sql
create or replace table ecommerce.gold.customer_lifetime_value as
select c.customer_unique_id,
c.customer_city,
c.customer_state,
sum(oi.price) as total_revenue,
sum(oi.freight_value) as total_freight,
avg(oi.price) as avg_order_value,
min(o.order_purchase_timestamp) as first_order,
max(o.order_purchase_timestamp) as latest_order

from ecommerce.silver.customers c
join ecommerce.silver.orders o on o.customer_id = c.customer_id
join ecommerce.silver.orderitems oi on oi.order_id = o.order_id
group by 1,2,3
```

```python
%sql
--Top Revenue Products
create or replace table ecommerce.gold.top_revenues
as 
select p.product_id ,p.product_category_name, sum(oi.price) as total_revenue,avg(oi.price) as avg_price, count(oi.order_id) as total_orders
from ecommerce.silver.products p 
join ecommerce.silver.orderitems oi on p.product_id = oi.product_id
group by 1,2
order by total_revenue desc;
```
